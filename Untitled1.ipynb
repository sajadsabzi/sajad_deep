```python
import cv2

cap = cv2.VideoCapture('1.mp4')  # ویدئوی ضبط‌شده توسط ربات

fgbg = cv2.createBackgroundSubtractorMOG2()
total_moving_pixels = 0
frame_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break
    fgmask = fgbg.apply(frame)
    moving_pixels = cv2.countNonZero(fgmask)
    total_moving_pixels += moving_pixels
    frame_count += 1

cap.release()

print('میانگین پیکسل‌های متحرک در هر فریم:', total_moving_pixels // frame_count)
```
