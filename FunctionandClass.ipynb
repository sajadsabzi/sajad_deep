```python
import math

def prime_numeric(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True
```

```python
def Collatzs_conjecture(n):
    if n < 1:
        print("Input must be a positive integer.")
        return
    sequence = [n]
    while n != 1: # تعداد تکرار را ۱۰ گذاشتیم
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        sequence.append(n)
    print(sequence)
```

```python
Collatzs_conjecture(13)
```

```python
def perfect_number(n):
    if n < 1:
        print("Input must be a positive integer.")
        return
    divisors = []
    for i in range(1, n):
        if n % i == 0:
            divisors.append(i)
    if sum(divisors) == n:
        print(f"{n} is a perfect number")
    else:
        print(f"{n} is not a perfect number")
```

```python
perfect_number(6)
```

```python
def Khayyam_Pascals_Triangle(n):
    triangle = []
    row = [1]
    for i in range(1, n):
        triangle.append(row.copy())  # کپی لیست
        row = [1] + [row[j] + row[j + 1] for j in range(len(row) - 1)] + [1]
    triangle.append(row.copy())
    return triangle
 
```

```python
import math
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def magnitude(self):
        return (self.x**2 + self.y**2)**0.5

    def direction(self):
        if self.x == 0 and self.y == 0:
            return None
        angle = math.atan2(self.y, self.x)
        return angle * (180 / math.pi)
```

```python
my_vector = Vector2D(3,4)
my_vector.magnitude()
my_vector.direction()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def __str__(self):
        return f"نام محصول {self.name} و قیمت آن {self.price} می باشد"

class ShoppingCart:
    def __init__(self):
        self.products = {}

    def add_product(self, product, quantity=1):
        if product.name in self.products:
            self.products[product.name]['quantity'] += quantity
        else:
            self.products[product.name] = {'product': product, 'quantity': quantity}

    def get_total(self):
        total = 0
        for item in self.products.values():
            total += item['product'].price * item['quantity']
        return total

    def __len__(self):
        return len(self.products.keys())

    def __str__(self):
        result = "نام محصول | تعداد | قیمت واحد | قیمت کل\n"
        result += "-" * 40 + "\n"
        for item in self.products.values():
            name = item['product'].name
            quantity = item['quantity']
            price = item['product'].price
            total_price = price * quantity
            result += f"{name:<12} | {quantity:^6} | {price:^10} | {total_price:^10}\n"
        result += "-" * 40 + "\n"
        result += f"جمع کل سبد خرید: {self.get_total()} تومان"
        return result
```

```python
P1 = Product("کتاب", 10000)
P2 = Product("خودکار", 2000)
P3 = Product("دفتر", 5000)
P4 = Product("پاکن", 2000)
cart = ShoppingCart()
cart.add_product(P1, 5)
cart.add_product(P2, 3)
cart.add_product(P3, 2)
cart.add_product(P4, 4)
#print(cart.__len__())
print(cart.__str__())
#print(cart.get_total())
```

```python
# فایل جدید بسازید مثلاً shopping_gui.py

import tkinter as tk
from tkinter import messagebox

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class ShoppingCart:
    def __init__(self):
        self.products = {}

    def add_product(self, product, quantity=1):
        if product.name in self.products:
            self.products[product.name]['quantity'] += quantity
        else:
            self.products[product.name] = {'product': product, 'quantity': quantity}

    def get_total(self):
        total = 0
        for item in self.products.values():
            total += item['product'].price * item['quantity']
        return total

    def get_table(self):
        result = "نام محصول | تعداد | قیمت واحد | قیمت کل\n"
        result += "-" * 40 + "\n"
        for item in self.products.values():
            name = item['product'].name
            quantity = item['quantity']
            price = item['product'].price
            total_price = price * quantity
            result += f"{name:<12} | {quantity:^6} | {price:^10} | {total_price:^10}\n"
        result += "-" * 40 + "\n"
        result += f"جمع کل سبد خرید: {self.get_total()} تومان"
        return result

# ساخت GUI
cart = ShoppingCart()

def add_to_cart():
    name = entry_name.get()
    try:
        price = int(entry_price.get())
        quantity = int(entry_quantity.get())
    except ValueError:
        messagebox.showerror("خطا", "قیمت و تعداد باید عدد باشند!")
        return
    product = Product(name, price)
    cart.add_product(product, quantity)
    text_result.delete(1.0, tk.END)
    text_result.insert(tk.END, cart.get_table())

root = tk.Tk()
root.title("سبد خرید ساده")

tk.Label(root, text="نام محصول:").grid(row=0, column=0)
entry_name = tk.Entry(root)
entry_name.grid(row=0, column=1)

tk.Label(root, text="قیمت واحد:").grid(row=1, column=0)
entry_price = tk.Entry(root)
entry_price.grid(row=1, column=1)

tk.Label(root, text="تعداد:").grid(row=2, column=0)
entry_quantity = tk.Entry(root)
entry_quantity.grid(row=2, column=1)

btn_add = tk.Button(root, text="افزودن به سبد خرید", command=add_to_cart)
btn_add.grid(row=3, column=0, columnspan=2)

text_result = tk.Text(root, width=50, height=10)
text_result.grid(row=4, column=0, columnspan=2)

root.mainloop()
```
