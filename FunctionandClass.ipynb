```python
import math

def prime_numeric(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True
```

```python
prime_numeric(20)
```

```python
def Collatzs_conjecture(n):
    if n < 1:
        print("Input must be a positive integer.")
        return
    sequence = [n]
    while n != 1: # تعداد تکرار را ۱۰ گذاشتیم
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        sequence.append(n)
    print(sequence)
```

```python
Collatzs_conjecture(13)
```

```python
def perfect_number(n):
    if n < 1:
        print("Input must be a positive integer.")
        return
    divisors = []
    for i in range(1, n):
        if n % i == 0:
            divisors.append(i)
    if sum(divisors) == n:
        print(f"{n} is a perfect number")
    else:
        print(f"{n} is not a perfect number")
```

```python
perfect_number(6)
```

```python
def Khayyam_Pascals_Triangle(n):
    triangle = []
    row = [1]
    for i in range(1, n):
        triangle.append(row.copy())  # کپی لیست
        row = [1] + [row[j] + row[j + 1] for j in range(len(row) - 1)] + [1]
    triangle.append(row.copy())
    return triangle
 
```

```python
Khayyam_Pascals_Triangle(6)
```
